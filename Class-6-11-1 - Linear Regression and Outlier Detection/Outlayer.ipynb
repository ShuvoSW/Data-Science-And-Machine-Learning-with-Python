```python
Q1 = df['height'].quantile(0.25)
Q3 = df.height.quantile(0.75)
```

```python
Q1, Q3
```

```python
IQR = Q3 - Q1
```

```python
IQR
```

```python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 - 1.5 * IQR
```

```python
lower_limit
```

```python
upper_limit
```

```python
df_no_outlier = df[(df.height >= lower_limit) & (df.height<=upper_limit)]
```
